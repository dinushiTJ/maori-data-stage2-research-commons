### Step 4: Data Cleaning - Handling Empty Columns in Hashed Dataset

As confirmed by `df.info()` and `df.describe()` for the hashed dataset, the 'record_id' and 'url' columns contain no data. We will drop these columns from the DataFrame to focus on relevant information.

```python
print('--- Dropping empty columns (record_id, url) from Hashed Dataset ---')
df_cleaned = df.drop(columns=['record_id', 'url'])
print('DataFrame after dropping columns:')
df_cleaned.info()
```

### Step 5: Initial Exploration of Object Columns in Hashed Dataset

Now that we've cleaned the DataFrame, let's look at the unique values and their counts for some of the key object-type columns to understand their distribution and potential for further analysis. This will include 'year', 'authors', 'subjects', and the new 'record_id_hash' column.

```python
print('\n--- Unique values and counts for \'year\' (top 20) ---')
display(df_cleaned['year'].value_counts().head(20))

print('\n--- Unique values and counts for \'authors\' (top 20) ---')
display(df_cleaned['authors'].value_counts().head(20))
```

```python
print('\n--- Unique values and counts for \'subjects\' (top 20) ---')
display(df_cleaned['subjects'].value_counts().head(20))
```

```python
print('\n--- Number of unique values for \'record_id_hash\' ---')
print(df_cleaned['record_id_hash'].nunique())

print('\n--- Checking for duplicate \'record_id_hash\' entries ---')
print(df_cleaned.duplicated(subset=['record_id_hash']).sum())
```

### Step 6: Further Data Cleaning - Handling Missing Values and Duplicates in Key Columns

Let's investigate the 'authors', 'abstract', and 'year' columns for missing values and duplicates to ensure data quality. We'll start by checking the count of null values in each of these columns.

```python
print('\n--- Missing values in \'authors\', \'abstract\', and \'year\' ---')
print(df_cleaned[['authors', 'abstract', 'year']].isnull().sum())
```

Now, let's examine potential duplicate entries. While duplicate authors or years are expected, a combination of title, abstract, and authors could indicate a duplicate record. We will check for duplicates based on these columns, excluding the unique `record_id_hash`.

```python
print('\n--- Checking for duplicate entries based on \'title\', \'abstract\', \'authors\', and \'year\' ---')
duplicate_entries = df_cleaned[df_cleaned.duplicated(subset=['title', 'abstract', 'authors', 'year'], keep=False)]

if not duplicate_entries.empty:
    print(f"Found {len(duplicate_entries)} potential duplicate entries.")
    display(duplicate_entries.sort_values(by=['title', 'year']).head())
else:
    print("No duplicate entries found based on 'title', 'abstract', 'authors', and 'year'.")
```

### Step 7: Analyzing Records with Same Author and Abstract but Different Years

Let's investigate if there are instances where the 'authors' and 'abstract' are identical, but the 'year' of publication differs. This could indicate revised versions of a paper, pre-prints, or just separate publications by the same authors on a very similar topic. We will group by 'authors' and 'abstract' and then check for multiple unique years within these groups.

```python
print('\n--- Identifying records with same authors and abstract but different years ---')

# Group by 'authors' and 'abstract' and count unique years
year_variations = df_cleaned.groupby(['authors', 'abstract'])['year'].nunique().reset_index()
year_variations.rename(columns={'year': 'unique_years_count'}, inplace=True)

# Filter for groups where the number of unique years is greater than 1
multi_year_entries_summary = year_variations[year_variations['unique_years_count'] > 1]

if not multi_year_entries_summary.empty:
    print(f"Found {len(multi_year_entries_summary)} groups with the same authors and abstract but multiple publication years.")
    display(multi_year_entries_summary)

    # Display all examples, instead of just the head(3)
    print('\n--- Examples of all such records ---')
    for index, row in multi_year_entries_summary.iterrows():
        authors = row['authors']
        abstract = row['abstract']
        print(f"\nAuthors: {authors}\nAbstract: {abstract[:100]}...") # Truncate abstract for display
        display(df_cleaned[(df_cleaned['authors'] == authors) & (df_cleaned['abstract'] == abstract)].sort_values(by='year'))
else:
    print("No records found with the same authors and abstract but different years.")
```

### Step 8: Detailed Review of Records with Same Authors & Abstract but Different Years

Let's perform a detailed review of these 18 groups to understand the nature of their differences, especially concerning the 'title' and 'subjects' columns, before deciding on a merging or cleaning strategy. This will help us determine if they are revised versions of the same work or distinct publications.

```python
print('\n--- Detailed Examination of Records with Year Variations ---')

if not multi_year_entries_summary.empty:
    for index, row in multi_year_entries_summary.iterrows():
        authors = row['authors']
        abstract = row['abstract']

        print(f"\nGroup {index + 1}:")
        print(f"  Authors: {authors}")
        print(f"  Abstract (truncated): {abstract[:150]}...") # Display a longer abstract snippet

        # Filter and display relevant columns for comparison within this group
        display_cols = ['title', 'year', 'subjects']
        display(df_cleaned[
            (df_cleaned['authors'] == authors) &
            (df_cleaned['abstract'] == abstract)
        ][display_cols].sort_values(by='year'))
else:
    print("No records found with the same authors and abstract but different years for detailed examination.")
```

```python
import difflib

print('\n--- Title Similarity Analysis for Multi-Year Groups ---')

if not multi_year_entries_summary.empty:
    for index, row in multi_year_entries_summary.iterrows():
        authors = row['authors']
        abstract = row['abstract']

        # Get all records for the current group
        group_records = df_cleaned[
            (df_cleaned['authors'] == authors) &
            (df_cleaned['abstract'] == abstract)
        ].sort_values(by='year').reset_index(drop=True)

        titles = group_records['title'].tolist()

        print(f"\nGroup {index + 1}: Authors: {authors}, Abstract: {abstract[:100]}...")
        print("  Titles:")
        for i, title in enumerate(titles):
            print(f"    [{i+1}] {title}")

        if len(titles) > 1:
            print("  Similarity Scores (Ratio):")
            # Calculate similarity for all unique pairs of titles
            for i in range(len(titles)):
                for j in range(i + 1, len(titles)):
                    title1 = titles[i]
                    title2 = titles[j]
                    similarity = difflib.SequenceMatcher(None, title1, title2).ratio()
                    print(f"    Title '{i+1}' vs Title '{j+1}': {similarity:.2f}")
        else:
            print("    (Only one title in this group, no similarity to calculate)")
else:
    print("No records found with the same authors and abstract but different years for similarity analysis.")
```

### Step 9: Handling Missing Values - Dropping Rows with Nulls in Key Columns

To ensure data quality for subsequent analysis, we will remove rows that have missing values in the `authors`, `abstract`, or `year` columns. This creates a new DataFrame `df_final` with complete records for these critical fields.

```python
print('--- Dropping rows with null values in authors, abstract, or year ---')
df_final = df_cleaned.dropna(subset=['authors', 'abstract', 'year']).copy()

print('\n--- DataFrame Information after dropping rows with nulls ---')
df_final.info()

print('\n--- Checking for remaining missing values in authors, abstract, and year ---')
print(df_final[['authors', 'abstract', 'year']].isnull().sum())
```

### Step 10: Re-checking for Duplicate Entries in `df_final`

Now that we've cleaned the `authors`, `abstract`, and `year` columns by dropping rows with null values, let's re-verify for duplicate entries based on the combination of `title`, `abstract`, `authors`, and `year` in our `df_final` DataFrame. This ensures data integrity after the cleaning process.

```python
print('\n--- Checking for duplicate entries in df_final based on \'title\', \'abstract\', \'authors\', and \'year\' ---')
duplicate_entries_final = df_final[df_final.duplicated(subset=['title', 'abstract', 'authors', 'year'], keep=False)]

if not duplicate_entries_final.empty:
    print(f"Found {len(duplicate_entries_final)} potential duplicate entries in df_final.")
    display(duplicate_entries_final.sort_values(by=['title', 'year']).head())
else:
    print("No duplicate entries found in df_final based on 'title', 'abstract', 'authors', and 'year'.")
```

### Step 11: Save the Cleaned DataFrame to CSV

Now that we have cleaned the data and handled missing values and duplicates in key columns, let's save the `df_final` DataFrame to a new CSV file. This will allow us to easily reuse the cleaned dataset for further analysis.

```python
num_rows = df_final.shape[0]
output_csv_path = f'/content/drive/MyDrive/NZ/University of Waikato/Māori Data Findability Project/stage2_colab/Datasets/6_cleaned_hashed_dataset_{num_rows}.csv'

try:
    df_final.to_csv(output_csv_path, index=False)
    print(f"Successfully saved the cleaned DataFrame to '{output_csv_path}'")
except Exception as e:
    print(f"An error occurred while saving the DataFrame: {e}")
```
